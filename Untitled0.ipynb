## Hyperparameter Tuning for Logistic Regression (Advanced)

```python
from sklearn.linear_model import LogisticRegression
from sklearn.model_selection import GridSearchCV
from sklearn.metrics import accuracy_score, classification_report

# Define a more extensive parameter grid for Logistic Regression
param_grid_lr = {
    'penalty': ['l1', 'l2', 'elasticnet'],  # Regularization type
    'C': [0.001, 0.01, 0.1, 1, 10, 100],  # Inverse of regularization strength
    'solver': ['liblinear', 'saga'],     # Solvers that support l1 and elasticnet
    'class_weight': [None, 'balanced']    # Handle class imbalance
}

# Initialize GridSearchCV for Logistic Regression
# Using the LogisticRegression model with random_state=42 for reproducibility
# and increasing max_iter to ensure convergence for complex solvers
grid_search_lr = GridSearchCV(estimator=LogisticRegression(random_state=42, max_iter=1000),
                              param_grid=param_grid_lr,
                              cv=5, # 5-fold cross-validation
                              n_jobs=-1, # Use all available cores
                              verbose=1, # Verbose output
                              scoring='accuracy') # Metric to optimize

# Fit GridSearchCV to the SMOTE-resampled training data
# Note: 'elasticnet' penalty requires 'saga' solver and l1_ratio parameter, which will be handled by GridSearchCV trying combinations.
# We will filter out invalid combinations if any, or rely on GridSearchCV to skip them.

# Filter out invalid combinations for specific solvers and penalties
# saga solver supports l1, l2, elasticnet
# liblinear solver supports l1, l2

valid_param_grid_lr = []
for p in param_grid_lr['penalty']:
    for c_val in param_grid_lr['C']:
        for solver_val in param_grid_lr['solver']:
            for cw in param_grid_lr['class_weight']:
                if p == 'elasticnet' and solver_val != 'saga':
                    continue # elasticnet only works with saga
                if p == 'l1' and solver_val not in ['liblinear', 'saga']:
                    continue # l1 works with liblinear, saga
                if p == 'l2' and solver_val not in ['lbfgs', 'newton-cg', 'liblinear', 'sag', 'saga']:
                    continue # l2 works with all
                valid_param_grid_lr.append({'penalty': [p], 'C': [c_val], 'solver': [solver_val], 'class_weight': [cw]})

# Re-initialize GridSearchCV with the filtered valid parameter combinations
grid_search_lr_filtered = GridSearchCV(estimator=LogisticRegression(random_state=42, max_iter=1000),
                                        param_grid=valid_param_grid_lr,
                                        cv=5, # 5-fold cross-validation
                                        n_jobs=-1, # Use all available cores
                                        verbose=1, # Verbose output
                                        scoring='accuracy') # Metric to optimize

grid_search_lr_filtered.fit(X_train_resampled, y_train_resampled)

# Print the best parameters and best score
print("\nBest parameters found for Logistic Regression: ", grid_search_lr_filtered.best_params_)
print("Best cross-validation accuracy for Logistic Regression: {:.4f}".format(grid_search_lr_filtered.best_score_))

# Get the best Logistic Regression model
best_lr_model = grid_search_lr_filtered.best_estimator_

# Evaluate the best Logistic Regression model on the original test set
y_pred_tuned_lr = best_lr_model.predict(X_test)
accuracy_tuned_lr = accuracy_score(y_test, y_pred_tuned_lr)

print(f"\nAccuracy of the best tuned Logistic Regression model on the test set: {accuracy_tuned_lr:.4f}")
print("\nClassification Report for the best tuned Logistic Regression model:")
print(classification_report(y_test, y_pred_tuned_lr))
```

## Feature Engineering and Preprocessing

```python
# Separate features (X) and target (y)
X = df_processed.drop('passed', axis=1)
y = df_processed['passed']

# Identify categorical and numerical features
categorical_features = X.select_dtypes(include='object').columns
numerical_features = X.select_dtypes(exclude='object').columns

print(f"Categorical Features: {list(categorical_features)}")
print(f"Numerical Features: {list(numerical_features)}")

# Apply One-Hot Encoding to categorical features
X_encoded = pd.get_dummies(X, columns=categorical_features, drop_first=True)

print("\nShape after One-Hot Encoding:", X_encoded.shape)
display(X_encoded.head())
```

## Data Splitting: Training and Testing Sets

```python
from sklearn.model_selection import train_test_split

# Split the data into training and testing sets
X_train, X_test, y_train, y_test = train_test_split(X_encoded, y, test_size=0.2, random_state=42)

print(f"X_train shape: {X_train.shape}")
print(f"X_test shape: {X_test.shape}")
print(f"y_train shape: {y_train.shape}")
print(f"y_test shape: {y_test.shape}")
```

## Model Training and Evaluation

```python
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score, classification_report

# Initialize and train a RandomForestClassifier model
model = RandomForestClassifier(random_state=42)
model.fit(X_train, y_train)

# Make predictions on the test set
y_pred = model.predict(X_test)

# Evaluate the model
accuracy = accuracy_score(y_test, y_pred)
print(f"Model Accuracy: {accuracy:.4f}")
print("\nClassification Report:")
print(classification_report(y_test, y_pred))
```

## Interpreting Model Performance

The accuracy score and classification report provide insights into how well our RandomForestClassifier performed. We can see metrics like precision, recall, and f1-score for both classes (passed/failed). While these give us a baseline, there's often room for improvement.

## Next Steps: Model Improvement through Hyperparameter Tuning

To enhance our model's predictive power, we can perform hyperparameter tuning. This involves trying different combinations of parameters for our `RandomForestClassifier` to find the set that yields the best performance on our validation data.

```python
from sklearn.model_selection import GridSearchCV

# Define the parameter grid to search
param_grid = {
    'n_estimators': [100, 200, 300],  # Number of trees in the forest
    'max_depth': [None, 10, 20, 30],  # Maximum depth of the tree
    'min_samples_split': [2, 5, 10],  # Minimum number of samples required to split an internal node
    'min_samples_leaf': [1, 2, 4]   # Minimum number of samples required to be at a leaf node
}

# Initialize GridSearchCV
grid_search = GridSearchCV(estimator=RandomForestClassifier(random_state=42),
                           param_grid=param_grid,
                           cv=5, # 5-fold cross-validation
                           n_jobs=-1, # Use all available cores
                           verbose=2, # Verbose output
                           scoring='accuracy') # Metric to optimize

# Fit GridSearchCV to the training data
grid_search.fit(X_train, y_train)

# Print the best parameters and best score
print("\nBest parameters found: ", grid_search.best_params_)
print("Best cross-validation accuracy: {:.4f}".format(grid_search.best_score_))

# Get the best model
best_model = grid_search.best_estimator_

# Evaluate the best model on the test set
y_pred_tuned = best_model.predict(X_test)
accuracy_tuned = accuracy_score(y_test, y_pred_tuned)

print(f"\nAccuracy of the best tuned model on the test set: {accuracy_tuned:.4f}")
print("\nClassification Report for the best tuned model:")
print(classification_report(y_test, y_pred_tuned))
```
